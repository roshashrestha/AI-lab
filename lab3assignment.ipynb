# Tensor
It is a multidimensional array used to store and process data. Everything including input data, model parameters, predictions and gradients in pytorch is represented as tensors. Basically, PyTorch performas all computations using tensors.


## Scalar
It is a single numerical value. It consists of magintude only and no dimensions.

```python
# creating scalar using tensor
scalar = torch.tensor(7)
scalar
```

torch.sensor(7) creates a tensor containing 7 values in which each of those values have no dimensions

```python
print(scalar.ndim) # ndim= no of dimensions
print(scalar.item()) #only works with one-element tensors
print(scalar.shape)
```

ndim gives 0 as scalar has no dimensions

item converts one element tenmsor into a standard python number so tensor(7) becomes 7

shape is for dimensions of tensor and scalar has no dimension so empty

## Vector
It is a 1D tensor which containes multiple values arranged in a single row(list).

```python
# vector
vector = torch.tensor([7, 8, 9, 10])
vector
```

creates 1D tensor with 4 elements

```python
print(vector.ndim)
print(vector.shape)
```

ndim gives 1 as it is 1D

tensor has 4 elements

You can tell the number of dimensions a tensor in PyTorch has by the number of square brackets on the outside `([)` and you only need to count one side.

**One of the most common errors in deep learning (shape errors)**

```python
# Shapes need to be in the right way  
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11], 
                         [9, 12]], dtype=torch.float32)

# torch.matmul(tensor_A, tensor_B) # (this will error)
```

```python
# The operation works when tensor_B is transposed
print(f"Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}\n")
print(f"New shapes: tensor_A = {tensor_A.shape} (same as above), tensor_B.T = {tensor_B.T.shape}\n")
print(f"Multiplying: {tensor_A.shape} * {tensor_B.T.shape} <- inner dimensions match\n")
print("Output:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output) 
print(f"\nOutput shape: {output.shape}")
```

**Reshaping**



```python
# Create a tensor
import torch
x = torch.arange(1., 9.)
print(x, x.shape)

# Add an extra dimension
x_reshaped = x.reshape(1, 8) # try other shapes, it will produce an error.
x_reshaped, x_reshaped.shape
```

```python
x, x.shape
```

**View**



```python
# Change view (keeps same data as original but changes view)
z = x.view(1, 8)
z, z.shape
```

```python
z = x.view(2,4)
z, z.shape
```

```python
x, x.shape
```

```python
import torch
x = torch.arange(6)
# print(x,x.shape)

a = x.view(2, 3)
b = x.reshape(2, 3)

a[0, 0] = 100
print(x)
```

That's the exact reason both functions exist: `view()` is a strict **"reinterpret the existing memory"** operation, while `reshape()` is a flexible **"give me this shape, even if a copy is needed"** operation.

**Stack**

- `torch.stack(tensors, dim=0)`:	Concatenates a sequence of tensors along a new dimension (dim), all tensors must be same size.

```python
# Stack tensors on top of each other
x_stacked = torch.stack([x, x, x, x], dim=0) 
x_stacked
```

```python
# Stacking using dim=1
x_stacked = torch.stack([x, x, x, x], dim=1) 
x_stacked
```

**Squeeze and Unsqueeze funtions.**

```python
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

# Remove extra dimension from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")


## Add an extra dimension with unsqueeze
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

**permute**



```python
import torch

x = torch.tensor([[1, 2, 3],
                  [4, 5, 6]])

print(x.shape)

y = x.permute(1, 0)

print(y)
print(y.shape)
```

## Indexing


```python
# Create a tensor 
import torch
x = torch.arange(1, 10).reshape(1, 3, 3)
x, x.shape

# Let's index bracket by bracket
print(f"First square bracket:\n{x[0]}") 
print(f"Second square bracket: {x[0][0]}") 
print(f"Third square bracket: {x[0][0][0]}")
```

You can also use : to specify "all values in this dimension" and then use a comma (,) to add another dimension.

```python
print(x[:, 0]) # Get all values of 0th dimension and the 0 index of 1st dimension
print(x[:, :, 1])# Get all values of 0th & 1st dimensions but only index 1 of 2nd dimension
print(x[:, 1, 1]) # Get all values of the 0 dimension but only the 1 index value of the 1st and 2nd dimension
print(x[0, 0, :]) # same as x[0][0]
```
